**Q1.** Write a Python program to take an integer as input and find the sum of its digits as output using a while loop.

```python
num = int(input("Enter a positive integer : "))
num2 = str(num)
sum = 0
count = 1
while count <= len(num2) :
  sum += int(num2[count-1])
  count += 1
print(f"The sum of the digits of {num} is",sum)
```

**Q2.** Write a Python program to take an integer as input and reverse the number using while loop.

```python
n = int(input("Enter a positive integer : "))
num = str(n)
alpha = ""
count = 1
while count <= len(num) :
  alpha += num[-count]
  count += 1
rev = int(alpha)
print(f"After reversing the digits of {n} we get",rev)
```

**Q3.** Write a Python program to repeatedly ask the user to enter a number until the user enters 0. At the end, show the sum of all the numbers entered.

```python
sum = 0
while True :
  ent = float(input("Enter a number : "))
  if ent == 0 :
    print("You have found the chosen number!")
    print("The sum of all the numbers you entered is",sum)
    break
  else :
    sum += ent
    print("Try again")


print("\nThank You.")
```
