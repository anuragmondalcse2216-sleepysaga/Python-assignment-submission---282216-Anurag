**TUPLES**

Q1. Remove Duplicate Elements :
Given a tuple containing duplicate values, create a new tuple containing only unique elements while preserving their original order.
Example : (1,2,2,3,1,4,3) → (1,2,3,4)

Q2. Sort of tuple of tuples :
Given students = (("Riya", 85), ("Aman", 92), ("Priya", 89))
Write a program to sort the students in descending order of marks and display their names and marks.

Q3. Write a Python program to create a tuple of 5 integers entered by the user. Calculate and display the sum and average of its elements.

Q4. Count Occurances :
Given a tuple, t = (10, 20, 10, 30, 10, 40, 20)
Write a program to count how many times the number 10 occurs in the tuple.

Q5. Find Maximum and Minimum :
Write a program to find the largest and smallest elements in a tuple without using the built-in max() and min() functions.

```python
#ANSWER 1

length = int(input("Enter size of tuple : "))

lst = []
for i in range(length) :
  x = int(input("Enter element : "))
  lst.append(x)
tup = tuple(lst)

newlst = []
for j in tup :
  if j not in newlst :
    newlst.append(j)
unq_tup = tuple(newlst)

print(f'The tuple {tup} with duplicate elements removed is {unq_tup}.')
```

```python
#ANSWER 2

students = (("Riya", 85), ("Aman", 92), ("Priya", 89))
lst_stud = list(students)
n = len(lst_stud)
for i in range(n) :
  for j in range(0 , n-i-1) :
    if lst_stud[j][1] < lst_stud[j+1][1] :
      lst_stud[j][1] , lst_stud[j+1][1] = lst_stud[j+1][1] , lst_stud[j][1]
sort_stud = tuple(lst_stud)

print(f'The tuple {students} sorted by the marks of the students is {sort_stud}.')
```

```python
#ANSWER 3
lst = []
sum = 0
for i in range(5) :
  val = int(input("Enter value of tuple : "))
  lst.append(val)
  sum += val
tup = tuple(lst)
print()
print(f'The tuple obtained is {tup}.')
avg = sum/5
print(f'The sum and average of all the elements in the tuple is {sum} and {avg} respectively.')
```

```python
#ANSWER 4

t = (10, 20, 10, 30, 10, 40, 20)
count = 0
for i in t :
  if i == 10 :
    count += 1

print(f'The value 10 occurs {count} times in {t}.')
```

```python
#ANSWER 5
n = int(input("Enter length of tuple : "))
lst = []
for i in range(n) :
  ele = int(input(f"Element {(i+1)} : "))
  lst.append(ele)
new = sorted(lst)
tup = tuple(new)
print("The tuple obtained is",tup)
print("\nThe greatest number in the list is",tup[n-1])
print("The smallest number in the list is",tup[0])
```

Q1. Given two sets, A = {1, 2, 3, 4, 5} and B = {4, 5, 6, 7, 8}, display their union, intersection, and difference (A - B).

Q2. Common Elements in Two Lists :
Write a program to accept two lists of integers and display all distinct elements that appear in both lists using sets.

Q3. Check Subset and Superset :
Given two sets, write a program to check whether the first set is a subset of the second set and whether the second set is a superset of the first.

Q4. Find Students Enrolled in Exactly One Course :
Two sets represent students enrolled in Python and Java courses.
python_students = {"Aman", "Riya", "Soham", "Priya"}
java_students = {"Riya", "Priya", "Rahul", "Ankit"}
Write a program to display students enrolled in exactly one of the two courses, excluding those enrolled in both.

```python
# ANSWER 1

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
uni = A.union(B)
inter = A.intersection(B)
diff = A.difference(B)

print(f'The union of the sets is {uni}.')
print(f'The intersection of the sets is {inter}.')
print(f'The difference of the sets is {diff}.')
```

```python
# ANSWER 3

len1 = int(input('Enter length of 1st list : '))
len2 = int(input('Enter length of 2nd list : '))
lst1 , lst2 = [],[]

for i in range(len1) :
  ele = int(input(f"Element {(i+1)} for 1st list : "))
  lst1.append(ele)

for i in range(len2) :
  ele = int(input(f"Element {(i+1)} for 2nd list : "))
  lst2.append(ele)

set1 = set(lst1)
set2 = set(lst2)

subch = set1.issubset(set2)
superch = set2.issuperset(set1)

print(f"Set A: {set1}")
print(f"Set B: {set2}\n")
print(f"Is Set A a subset of Set B? {subch}")
print(f"Is Set B a superset of Set A? {superch}")

```

**DICTIONARY**

Q1. Student Marks Dictionary :
Create a dictionary containing the names and marks of 5 students. Display the name and marks of each student using a loop.

Q2. Frequency of Words :
Write a Python program to accept a sentence and create a dictionary containing the frequency of each word.
Example: "python is easy and python is useful"
Expected result: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'useful': 1}

Q3. Merge Two Dictionaries and Add Common Values :
Given two dictionaries containing item names and quantities, merge them into a single dictionary. If an item appears in both dictionaries, add its quantities.
Example :
d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}
Expected result : {"Pen": 14, "Pencil": 12, "Book": 3, "Eraser": 6}


```python
# ANSWER 1

dict = {}
for i in range(5) :
  name = input("Enter name : ")
  marks = float(input("Enter marks : "))
  dict[name] = marks
for i in dict :
  print (f'{i} scored {dict[i]}.')
```

```python
# ANSWER 2

sentence = input("Enter a sentence : ")
words = sentence.split()
freq = {}
for word in words :
  word = word.lower()
  if word in freq:
    freq[word] += 1
  else:
    freq[word] = 1
print("Expected result:",freq)
```

```python
# ANSWER 3

d1,d2 = {},{}
len1 = int(input('Enter length of 1st dictionary : '))
len2 = int(input('Enter length of 2nd dictionary : '))
for i in range(len1) :
  item = input("Enter item for 1st dictionary : ")
  qty = int(input("Enter quantity : "))
  d1[item] = qty
for i in range(len2) :
  item = input("Enter item 2nd dictionary : ")
  qty = int(input("Enter quantity : "))
  d2[item] = qty
result = d1.copy()
for item,qty in d2.items():
   result[item] = result.get(item, 0) + qty
print(result)
```
