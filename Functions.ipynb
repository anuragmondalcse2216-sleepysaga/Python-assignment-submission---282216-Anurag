**Q1.** Write a Python program with separate functions for addition, subtraction, multiplication, and division. Take two numbers from the user and display the result.

```python
def add(a,b) :
  r = a+b
  print(f'{a} + {b} = {r}')

def sub(a,b) :
  r = a-b
  print(f'{a} - {b} = {r}')

def mul(a,b) :
  r = a*b
  print(f'{a} × {b} = {r}')

def div(a,b) :
  r = a/b
  print(f'{a} ÷ {b} = {r}')

print("----- BASIC CALCULATOR -----")
print()
print('''Math operation menu :
1. ADDITION (+)
2. SUBTRACTION (-)
3. MULTIPLICATION (×)
4. DIVISION (÷)''')
print()

while True :
  num1 = float(input("Enter 1st number : "))
  num2 = float(input("Enter 2nd number : "))
  op = input("Enter operation (1/2/3/4) : ")
  if op=='1' :
    add(num1,num2)
  elif op=='2' :
    sub(num1,num2)
  elif op=='3' :
    mul(num1,num2)
  elif op=='4' :
    div(num1,num2)
  que = input("Try again? (y/n) : ")
  if que=='n' :
    print("Thank You")
    break
```

**Q2.** Write a Python program to create functions to convert : Celcius → Farenheit, Farenheit → Celcius, Celcius → Kelvin.

```python
def celtofar() :
  cel = float(input("Enter temperature in celcius : "))
  far = (cel*9/5)+32
  print(f'{cel}℃ is equal to {far}℉.')

def fartocel() :
  far = float(input("Enter temperature in farenheit : "))
  cel = (far-32)*5/9
  print(f'{far}℉ is equal to {cel}℃.')

def celtokel() :
  cel = float(input("Enter temperature in celcius : "))
  kel = cel+273.15
  print(f'{cel}℃ is equal to {kel}K.')

celtofar()
fartocel()
celtokel()
```

**Q4.** Write a Python program to create a function that takes a number as a parameter and checks whether the number is prime or not.

```python
def primecheck(num) :
  if num == 2:
    print(f"The number {num} is a prime number.")
  elif num > 0:
    for i in range(2,num) :
      if num%i == 0:
        check = True
        break
      else :
        check = False
    if check == True :
      print(f"The number {num} is NOT a prime number.")
    elif check == False :
      print(f"The number {num} is a prime number.")
  else :
    print(f"The number {num} is NOT a prime number.")

x = int(input("Enter a positive integer : "))
primecheck(x)
```

**Q5.** Write a Python program to reverse a string (Define a function named 'string_reverse').

```python
def string_reverse() :
  str = input('Enter a word or string : ')
  rev = str[::-1]
  print(f'The reverse of {str} is {rev}.')

string_reverse()
```
