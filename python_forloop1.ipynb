**Q1.** Write a Python program to print all the numbers between 1 and 100 that are divisible by both 3 and 5.

```python
print("The numbers that are divisible by both 3 and 5 from 1 to 100 : ")
for i in range(1,101) :
  if i%(3*5) == 0 :
    print(i)
```

**Q2.** Write a Python program to take an integer n as an input and calculate the sum of the following series :
1²+2²+3²+4²+...+n² = ?

```python
print('For the sum of the series : 1²,2²,3²,4²,...,n²\n')
n = int(input("Choose a positive integer : "))
sum=0
if n>0 :
  for i in range(1,n+1) :
    sum+=(i**2)
  print(f"The sum of series for {n} is {sum}.")
else :
  print("Please enter a positive integer number!")
```

**Q3.** Write a Python program to take a number as input and check whether it is a prime number.

```python
num = int(input('Choose an integer : '))
if num == 2:
  print(f"The number {num} is a prime number.")
elif num > 0 :
  for i in range(2,num) :
    if num%i == 0:
      check = True
      break
    else :
      check = False
  if check == True :
    print(f"The number {num} is NOT a prime number.")
  elif check == False :
    print(f"The number {num} is a prime number.")
else :
  print(f"The number {num} is NOT a prime number.")
```
