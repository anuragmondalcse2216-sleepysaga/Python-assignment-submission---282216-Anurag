**Q1.** Write a Python program to find sum and average of any of the first 55 natural numbers.

```python
nat = int(input("Enter a number between 1 and 55 : "))

while nat in range(1,56) :
  add = nat*(nat+1)/2
  avg = add/nat
  print( f"\nThe sum of the first {nat} natural numbers is {add}" )
  print( f"The average of the first {nat} natural numbers is {avg}." )
  que = input("\nDo you want to repeat? (y/n) : ")
  if que == 'n' or que != 'y' :
    print('\nTask complete!')
    break
  elif que == 'y' :
    nat = int(input("\nEnter a number between 1 and 55 : "))
else:
  print("\nThe number you chose is out of the given range!")
```

**Q2.** Write a Python program to find the factorial of any number till 7 with a loop.

```python
num = int(input("Enter a positive number till 7 : "))
fact = 1

while num in range(8) :
  if num == 0 :
    print('The factorial of 0 is 1.')
  else :
    for i in range(1,num+1) :
      fact = fact*i
    print( f"The factorial of {num} is {fact}." )
  que = input("\nDo you want to repeat? (y/n) : ")
  if que == 'n' or que != 'y' :
    print('\nTask complete!')
    break
  elif que == 'y' :
    num = int(input("\nEnter a positive number till 7 : "))
    fact = 1
else :
  print('\nInput is out of range!')
```

**Q3.** Write a Python program that takes a string of characters and reverses it.

```python
st = input("Enter a string of characters : ")

while True :
  rev = st[::-1]
  print( f"The reverse of {st} is {rev}." )
  que = input("\nDo you want to try again? (y/n) : ")
  if que == 'n' or que != 'y' :
    print('\nTask complete!')
    break
  elif que == 'y' :
    st = input("\nEnter a string of characters : ")
```

**Q4.** Write a Python program which can check whether a given year is a leap year or not.

```python
year = int(input("Give an year (in digits) to check : "))
while True :
  if year%4 == 0 :
    print( f"The year {year} is a leap year." )
  else :
    print( f"The year {year} is NOT a leap year." )
  que = input("\nDo you want to try again? (y/n) : ")
  if que == 'n' or que != 'y' :
    print('\nTask complete!')
    break
  elif que == 'y' :
    year = int(input("\nGive an year (in digits) to check : "))
```
