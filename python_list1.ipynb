**Q1.** Write a Python program to take 10 numbers from the user and store them in a list. Then find the show the greatest and the smallest elements without using 'max()' and 'min()' functions.



```python
print("Enter 10 numbers to add in the list →")
lst = []
for i in range(10) :
  ele = float(input(f"Element {(i+1)} : "))
  lst.append(ele)
new = lst.sort()
print("The list obtained is",lst)
print("\nThe greatest number in the list is",new[9])
print("The smallest number in the list is",new[0])
```

**Q2.** Write a Python program to remove all duplicate elements from a list without using 'set()' function.

```python
len = int(input("Enter number of elements to be added in list : "))
lst = []
for i in range(1,len+1) :
  ele = eval(input(f"Element no. {i} : "))
  lst.append(ele)
print(f"\nThe obtained list is {lst}")
new_lst = []
for j in lst :
  if j not in new_lst :
    new_lst.append(j)
lst = new_lst
print(f"The list with no duplicate values is {lst}")
```
